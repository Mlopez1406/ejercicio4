```python
'''
Algoritmo: Signo de un número
Entrada: numero (numérico)
salida: cadena de texto ('positivo', 'negativo' o 'cero')
bloque de desarrollo:
  Si numero > 0 entonces
    devolver 'positivo'
  Sino, si numero < 0 entonces
    devolver 'negativo'
  Sino
    devolver 'cero'
  Fin si
'''
#Algoritmo Signo de un número en python
def signo_numero(numero):
    """Devuelve si un número es positivo, negativo o cero."""
    if numero > 0:
        return 'positivo'
    elif numero < 0:
        return 'negativo'
    else:
        return 'cero'

# Ejemplos de uso para probar la celda:
print(signo_numero(-5))
print(signo_numero(0))
```

```python
'''
Algoritmo: Año bisiesto
Entrada: anio (entero)
salida: booleano (True o False)
bloque de desarrollo:
  Si anio es divisible por 4 y (no es divisible por 100 o es divisible por 400) entonces
    devolver True
  Sino
    devolver False
  Fin si
'''
#Algoritmo Año bisiesto en python
def es_bisiesto(anio):
    """Determina si un año es bisiesto."""
    if anio % 4 == 0 and (anio % 100 != 0 or anio % 400 == 0):
        return True
    return False

# Ejemplos de uso para probar la celda:
print(es_bisiesto(2024))
print(es_bisiesto(1900))
print(es_bisiesto(2000))
```

```python
'''
Algoritmo: Mayor de tres números
Entrada: a, b, c (numéricos)
salida: numérico (el mayor)
bloque de desarrollo:
  Si a >= b y a >= c entonces
    devolver a
  Sino, si b >= a y b >= c entonces
    devolver b
  Sino
    devolver c
  Fin si
'''
#Algoritmo Mayor de tres números en python
def mayor_de_tres(a, b, c):
    """Devuelve el mayor de tres números sin usar max()."""
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

# Ejemplos de uso para probar la celda:
print(mayor_de_tres(4, 9, 2))
print(mayor_de_tres(7, 7, 3))
```

```python
'''
Algoritmo: Categoría de envío
Entrada: peso_kg (numérico)
salida: cadena de texto (categoría)
bloque de desarrollo:
  Si peso_kg <= 1 entonces
    devolver 'Sobre'
  Sino, si peso_kg <= 5 entonces
    devolver 'Paquete pequeño'
  Sino, si peso_kg <= 20 entonces
    devolver 'Paquete grande'
  Sino
    devolver 'Voluminoso'
  Fin si
'''
#Algoritmo Categoría de envío en python
def categoria_envio(peso_kg):
    """Clasifica un envío según su peso en kilos."""
    if peso_kg <= 1:
        return 'Sobre'
    elif peso_kg <= 5:
        return 'Paquete pequeño'
    elif peso_kg <= 20:
        return 'Paquete grande'
    else:
        return 'Voluminoso'

# Ejemplos de uso para probar la celda:
print(categoria_envio(0.5))
print(categoria_envio(15))
```

```python
'''
Algoritmo: Suma de pares
Entrada: n (entero)
salida: entero (suma)
bloque de desarrollo:
  suma = 0
  Para i desde 1 hasta n hacer
    Si i es par entonces
      suma = suma + i
    Fin si
  Fin para
  devolver suma
'''
#Algoritmo Suma de pares en python
def suma_pares(n):
    """Suma los números pares entre 1 y n."""
    suma = 0
    for i in range(1, n + 1):
        if i % 2 == 0:
            suma += i
    return suma

# Ejemplos de uso para probar la celda:
print(suma_pares(10))
print(suma_pares(1))
```

```python
'''
Algoritmo: Factorial iterativo
Entrada: n (entero)
salida: entero (factorial)
bloque de desarrollo:
  resultado = 1
  Para i desde 1 hasta n hacer
    resultado = resultado * i
  Fin para
  devolver resultado
'''
#Algoritmo Factorial iterativo en python
def factorial_iterativo(n):
    """Calcula n! usando un bucle (versión iterativa)."""
    resultado = 1
    for i in range(1, n + 1):
        resultado *= i
    return resultado

# Ejemplos de uso para probar la celda:
print(factorial_iterativo(5))
print(factorial_iterativo(0))
```

```python
'''
Algoritmo: Contar vocales
Entrada: palabra (cadena)
salida: entero (cantidad de vocales)
bloque de desarrollo:
  contador = 0
  Para cada letra en palabra hacer
    Si letra en minúscula es 'a', 'e', 'i', 'o', 'u' entonces
      contador = contador + 1
    Fin si
  Fin para
  devolver contador
'''
#Algoritmo Contar vocales en python
def contar_vocales(palabra):
    """Cuenta las vocales de una palabra, sin distinguir mayúsculas."""
    contador = 0
    vocales = 'aeiou'
    for letra in palabra.lower():
        if letra in vocales:
            contador += 1
    return contador

# Ejemplos de uso para probar la celda:
print(contar_vocales('Programacion'))
print(contar_vocales('xyz'))
```

```python
'''
Algoritmo: Tabla de multiplicar
Entrada: numero (entero)
salida: ninguna (imprime por pantalla)
bloque de desarrollo:
  Para i desde 1 hasta 10 hacer
    imprimir numero + "x" + i + "=" + (numero * i)
  Fin para
'''
#Algoritmo Tabla de multiplicar en python
def tabla_multiplicar(numero):
    """Imprime la tabla de multiplicar de un número, del 1 al 10."""
    for i in range(1, 11):
        print(f"{numero}x{i}={numero * i}")

# Ejemplo de uso para probar la celda:
tabla_multiplicar(3)
```

```python
'''
Algoritmo: Calcular promedio
Entrada: numeros (lista de numéricos)
salida: numérico (promedio)
bloque de desarrollo:
  Si la lista está vacía entonces devolver 0.0
  suma = 0
  Para cada num en numeros hacer
    suma = suma + num
  Fin para
  devolver suma / longitud(numeros)
'''
#Algoritmo Calcular promedio en python
def calcular_promedio(numeros):
    """Calcula la media aritmética de una lista de números."""
    if not numeros:
        return 0.0
    return sum(numeros) / len(numeros)

# Ejemplos de uso para probar la celda:
print(calcular_promedio([4, 8, 6]))
```

```python
'''
Algoritmo: Analizar frase
Entrada: frase (cadena)
salida: tupla (cadena en mayúsculas, entero número de palabras)
bloque de desarrollo:
  frase_mayus = convertir_a_mayusculas(frase)
  palabras = dividir_por_espacios(frase)
  numero_palabras = longitud(palabras)
  devolver frase_mayus, numero_palabras
'''
#Algoritmo Analizar frase en python
def analizar_frase(frase):
    """Devuelve la frase en mayúsculas junto con su número de palabras."""
    return frase.upper(), len(frase.split())

# Ejemplos de uso para probar la celda:
print(analizar_frase('hola que tal'))
```

```python
'''
Algoritmo: Validar contraseña
Entrada: contrasena (cadena), longitud_minima (entero, por defecto 8)
salida: booleano
bloque de desarrollo:
  Si longitud(contrasena) >= longitud_minima entonces
    devolver Verdadero
  Sino
    devolver Falso
  Fin si
'''
#Algoritmo Validar contraseña en python
def validar_contrasena(contrasena, longitud_minima=8):
    """Comprueba si una contraseña alcanza la longitud mínima exigida."""
    return len(contrasena) >= longitud_minima

# Ejemplos de uso para probar la celda:
print(validar_contrasena('abc123'))
print(validar_contrasena('abc123', 5))
```

```python
'''
Algoritmo: Celsius a Fahrenheit y Fahrenheit a Celsius
Entrada: grados (numérico)
salida: numérico (grados convertidos)
bloque de desarrollo (Celsius a Fahrenheit):
  devolver (celsius * 9/5) + 32
bloque de desarrollo (Fahrenheit a Celsius):
  devolver (fahrenheit - 32) * 5/9
'''
#Algoritmo Conversión de temperatura en python
def celsius_a_fahrenheit(celsius):
    """Convierte una temperatura de Celsius a Fahrenheit."""
    return (celsius * 9/5) + 32

def fahrenheit_a_celsius(fahrenheit):
    """Convierte una temperatura de Fahrenheit a Celsius."""
    return (fahrenheit - 32) * 5/9

# Ejemplos de uso para probar la celda:
print(celsius_a_fahrenheit(0))
print(fahrenheit_a_celsius(32))
```

```python
'''
Algoritmo: Factorial recursivo
Entrada: n (entero)
salida: entero
bloque de desarrollo:
  Si n == 0 o n == 1 entonces
    devolver 1
  Sino
    devolver n * factorial_recursivo(n - 1)
  Fin si
'''
#Algoritmo Factorial recursivo en python
def factorial_recursivo(n):
    """Calcula n! llamándose a sí misma (versión recursiva)."""
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursivo(n - 1)

# Ejemplos de uso para probar la celda:
print(factorial_recursivo(5))
print(factorial_recursivo(0))
```

```python
'''
Algoritmo: Fibonacci
Entrada: n (entero)
salida: entero
bloque de desarrollo:
  Si n == 0 entonces devolver 0
  Si n == 1 entonces devolver 1
  devolver fibonacci(n - 1) + fibonacci(n - 2)
'''
#Algoritmo Fibonacci en python
def fibonacci(n):
    """Devuelve el término n de la sucesión de Fibonacci."""
    if n == 0:
        return 0
    elif n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

# Ejemplos de uso para probar la celda:
print(fibonacci(6))
print(fibonacci(0))
```

```python
'''
Algoritmo: Suma lista recursiva
Entrada: numeros (lista)
salida: numérico
bloque de desarrollo:
  Si la lista está vacía entonces devolver 0
  devolver primer elemento + suma_lista(resto de los elementos)
'''
#Algoritmo Suma lista recursiva en python
def suma_lista(numeros):
    """Suma los elementos de una lista de forma recursiva."""
    if not numeros:
        return 0
    return numeros[0] + suma_lista(numeros[1:])

# Ejemplos de uso para probar la celda:
print(suma_lista([1, 2, 3, 4]))
print(suma_lista([]))
```

```python
'''
Algoritmo: Invertir texto recursivo
Entrada: texto (cadena)
salida: cadena
bloque de desarrollo:
  Si longitud(texto) <= 1 entonces devolver texto
  devolver último caracter + invertir_texto(texto sin el último caracter)
'''
#Algoritmo Invertir texto recursivo en python
def invertir_texto(texto):
    """Invierte una cadena de texto de forma recursiva."""
    if len(texto) <= 1:
        return texto
    return texto[-1] + invertir_texto(texto[:-1])

# Ejemplos de uso para probar la celda:
print(invertir_texto('python'))
print(invertir_texto('a'))
```

```python
'''
Algoritmo: Convertir a entero
Entrada: texto (cadena)
salida: entero o cadena de texto (error)
bloque de desarrollo:
  Intentar:
    devolver texto convertido a entero
  Capturar error de valor:
    devolver 'Error: no es un número'
'''
#Algoritmo Convertir a entero en python
def convertir_a_entero(texto):
    """Convierte un texto a entero controlando el error de formato."""
    try:
        return int(texto)
    except ValueError:
        return 'Error: no es un número'

# Ejemplos de uso para probar la celda:
print(convertir_a_entero('42'))
print(convertir_a_entero('hola'))
```

```python
'''
Algoritmo: Obtener elemento
Entrada: lista (lista), posicion (entero)
salida: elemento o cadena de texto (error)
bloque de desarrollo:
  Intentar:
    devolver lista[posicion]
  Capturar error de índice:
    devolver 'Error: posición fuera de rango'
'''
#Algoritmo Obtener elemento en python
def obtener_elemento(lista, posicion):
    """Accede a una posición de una lista controlando el error de rango."""
    try:
        return lista[posicion]
    except IndexError:
        return 'Error: posición fuera de rango'

# Ejemplos de uso para probar la celda:
print(obtener_elemento([1, 2, 3], 1))
print(obtener_elemento([1, 2, 3], 9))
```

```python
'''
Algoritmo: Calculadora segura
Entrada: texto_a, texto_b (cadenas), operacion (cadena)
salida: numérico o cadena de texto (error)
bloque de desarrollo:
  Intentar:
    a = convertir a flotante(texto_a)
    b = convertir a flotante(texto_b)
    Si operacion es 'suma' entonces
      devolver a + b
    Sino si operacion es 'division' entonces
      devolver a / b
  Capturar error de valor:
    devolver 'Error: valor no numérico'
  Capturar error de división por cero:
    devolver 'Error: división entre cero'
'''
#Algoritmo Calculadora segura en python
def calculadora_segura(texto_a, texto_b, operacion):
    """Suma o divide dos textos numéricos controlando los errores posibles."""
    try:
        a = float(texto_a)
        b = float(texto_b)
        if operacion == 'suma':
            return a + b
        elif operacion == 'division':
            return a / b
    except ValueError:
        return 'Error: valor no numérico'
    except ZeroDivisionError:
        return 'Error: división entre cero'

# Ejemplos de uso para probar la celda:
print(calculadora_segura('10', '0', 'division'))
print(calculadora_segura('abc', '2', 'suma'))
print(calculadora_segura('4', '2', 'suma'))
```
